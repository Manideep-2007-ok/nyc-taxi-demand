```python
import pandas as pd
jan = pd.read_parquet("https://d37ci6vzurychx.cloudfront.net/trip-data/yellow_tripdata_2025-01.parquet")
print(jan.shape)
print(jan.columns)
```

```python
jan["hour"] = jan["tpep_pickup_datetime"].dt.hour
print(jan["hour"].head())
```

```python
jan["hour_start"] = jan["tpep_pickup_datetime"].dt.floor("h")
jan_clean = jan[(jan["tpep_pickup_datetime"]>="2025-01-01")&(jan["tpep_pickup_datetime"]<"2025-02-01")]
```

```python
print(jan_clean.groupby(["hour_start","PULocationID"]).size())
```
